```python
import pandas as pd
import numpy as np

```

```python
np.random.seed(42) # keeps your numbers the same every time you re-run this

months = pd.date_range(start='2024-01-01', periods=24 , freq='MS')

#Simulate a growing startup: customers grow, revenue grows, expenses  grow slower
customers = np.round(50 * (1.15 ** np.arange(24))) .astype(int)
avg_revenue_per_customer = np.random.normal(120,10,24)
revenue = customers * avg_revenue_per_customer
expenses = revenue * np.random.uniform(0.7,0.95,24)

df_financials = pd.DataFrame({
    'Month': months,
    'Customers': customers,
    'Revenue': revenue.round(2),
    'Expenses': expenses.round(2)
})

df_financials.head()
```

```python
# Expenses start high relative to revenue (buring cash) and improve over time
burn_multiplier = np.linspace(1.6,1.15,24)
expenses = revenue * burn_multiplier * np.random.uniform(0.95,1.05,24)

df_financials['Expenses'] = expenses.round(2)
df_financials['Net_Burn'] = df_financials['Expenses'] - df_financials['Revenue']
df_financials['Cash_Balance'] = 500000 - df_financials['Net_Burn'].cumsum()

df_financials[['Month','Revenue','Expenses','Net_Burn','Cash_Balance']].head()
```

```python
# New customers each month (difference from previous month)
df_financials['New_Customers'] = df_financials['Customers'].diff().fillna(df_financials['Customers'][0]).astype(int)

# Simulate marketing spend - realsitic: roughly 20-30% of revenue goes to marketing early on
df_financials['Marketing_Spend'] =  (df_financials['Revenue'] * np.random.uniform(0.2, 0.3, 24)).round(2)

#CAC = marketing spend/new customers acquired that much
df_financials['CAC'] = (df_financials['Marketing_Spend']/ df_financials['New_Customers']).round(2)

df_financials[['Month','New_Customers','Marketing_Spend','CAC']].head(10)


```

```python
#ARPU = Average Revenue Per User (per customer,per month)
df_financials['ARPU'] = (df_financials['Revenue'] / df_financials['Customers']).round(2)

# Simulate a realistic monthly churn rate for an early-stage startup(typically 4-7%)
np.random.seed(7)
monthly_churn_rate = np.random.uniform(0.04,0.07,24)
df_financials['Churn_Rate'] = (monthly_churn_rate * 100).round(2)

# Average customer lifetime (in months) = 1 / churn rate
df_financials['Avg_Customer_Lifetime_Months'] = (1/monthly_churn_rate).round(1)

# LTV = ARPU * average lifetime
df_financials['LTV'] = (df_financials['ARPU'] * df_financials['Avg_Customer_Lifetime_Months']).round(2)

# The single most important ratio in this whole project
df_financials['LTV_CAC_Ratio'] = (df_financials['LTV'] / df_financials['CAC']).round(2)

df_financials[['Month','ARPU','Churn_Rate','LTV','CAC','LTV_CAC_Ratio']].head(10)

```

```python
# Average churn rate across the whole period (simplification for a clean , explainable cohort table)
avg_churn = monthly_churn_rate.mean()
print(f"Average monthly churn rate: {avg_churn*100:.2f}%")

# Build a cohort retention table:
# rows = the month a customer group joined("Cohort")
# Columns = how many months since they joined
# values = % of that cohort still active

max_months_tracked = 12
cohort_table = pd.DataFrame(index=df_financials['Month'][:12], columns= range(max_months_tracked))

for i , cohort_month in enumerate(df_financials['Month'][:12]):
  for months_since_joining in range (max_months_tracked):
    #Retention formula: what % survives after N months of churn compounding
    retention_pct = ((1-avg_churn) ** months_since_joining) * 100
    cohort_table.loc[cohort_month, months_since_joining] = round(retention_pct,1)

cohort_table
```

```python
# Visualize it as a heatmap - this is the classic "cohort analysis" chart you'll see everywhere

import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))
sns.heatmap(cohort_table.astype(float), annot=True , fmt='.1f' , cmap='YlOrRd')
plt.title('Customer Retention by Cohort (%)')
plt.xlabel('Months Since Joining')
plt.ylabel('Cohort( Signup Month)')
plt.show()
```

```python
# Export both tables to a single Excel file with two sheets
with pd.ExcelWriter('financial_kpi_data.xlsx', engine = 'openpyxl') as writer:
  df_financials.to_excel(writer, sheet_name='Monthly_Financials', index= False)
  cohort_table.to_excel(writer, sheet_name='Cohort_Retention')

from google.colab import files
files.download('financial_kpi_data.xlsx')
```
